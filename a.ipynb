```python
import pandas as pd
```

```python
messi = pd.read_csv("eventsMessi.csv")
```

```python
goals = messi[messi['shot_outcome']=='Goal']
```

```python
goals
```

```python
goalMID = goals['match_id'].tolist()
```

```python
mMatches = pd.read_csv("messiMatches.csv")
```

```python
goalDate = pd.DataFrame(columns=['goalNumber', 'Date'])
i=len(goalMID)
for mid in goalMID:
    for index,row in mMatches.iterrows():
        if(row['match_id'] == mid):
            goalDate.loc[len(goalDate)] = [i,row['match_date']]
            i=i-1
    
```

```python
goalDate
```

```python
assistMID = messi[messi['pass_goal_assist']==True]['match_id'].tolist()
```

```python
assistMID
```

```python
assistDate = pd.DataFrame(columns=['assistNumber', 'Date'])
i=len(assistMID)
for mid in assistMID:
    for index,row in mMatches.iterrows():
        if(row['match_id'] == mid):
            assistDate.loc[len(assistDate)] = [i,row['match_date']]
            i=i-1
```
